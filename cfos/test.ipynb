```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# 1. 데이터 생성 (50행 x 1000열)
data = np.random.rand(50, 1000)

# 2. 캔버스 설정
plt.figure(figsize=(12, 6))

# 3. 히트맵 그리기
# xticklabels=False로 설정하여 1~1000까지의 지저분한 기본 인덱스를 숨깁니다.
ax = sns.heatmap(data, xticklabels=False)

# 4. 범위를 지정하고 라벨 설정하기
# 예: (시작 인덱스, 끝 인덱스, "표시할 이름")
ranges = [
    (0, 122, "Front, Motor, Soma, Vis, Gust"),
    (123, 700, "Group B (300-700)"),
    (700, 1000, "Group C (700-1000)")
]

# 라벨이 찍릴 위치(ticks)와 이름(labels)을 저장할 리스트
xticks_positions = []
xticklabels_names = []

for start, end, label in ranges:
    # 범위의 '중심' 위치 계산
    center = (start + end) / 2
    xticks_positions.append(center)
    xticklabels_names.append(label)
    
    # (선택사항) 구분선 그리기: 범위의 끝지점에 선을 그어 구역을 나눔
    if end < 1000:
        plt.axvline(x=end, color='white', linestyle='--', linewidth=1)

# 5. x축 틱과 라벨 적용
ax.set_xticks(xticks_positions)
ax.set_xticklabels(xticklabels_names, rotation=0, fontsize=12)

# x축 제목 (선택)
plt.xlabel("Index Ranges")

plt.show()
```
